```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder
```

```python
df = pd.read_csv("../data/raw/sl_restaurant_demand_raw.csv")

df.head()
```

```python
df.info()
df.isna().sum()
```

```python
df["date"] = pd.to_datetime(df["date"])
```

```python
label_encoders = {}

categorical_cols = [
    "day_of_week",
    "menu_category",
    "festival_type"
]

for col in categorical_cols:
    le = LabelEncoder()
    df[col] = le.fit_transform(df[col])
    label_encoders[col] = le
```

```python
df = df.sort_values("date").reset_index(drop=True)
```

```python
X = df.drop(columns=["daily_demand_qty", "date"])
y = df["daily_demand_qty"]
```

```python
split_index = int(len(df) * 0.8)

X_train = X.iloc[:split_index]
X_test  = X.iloc[split_index:]

y_train = y.iloc[:split_index]
y_test  = y.iloc[split_index:]
```

```python
train_df = X_train.copy()
train_df["daily_demand_qty"] = y_train

test_df = X_test.copy()
test_df["daily_demand_qty"] = y_test

train_df.to_csv("../data/processed/train.csv", index=False)
test_df.to_csv("../data/processed/test.csv", index=False)

print("✅ Preprocessed train & test datasets saved")
```
