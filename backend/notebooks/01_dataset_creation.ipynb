```python
import pandas as pd
import numpy as np
from datetime import datetime
```

```python
np.random.seed(42)

START_DATE = "2023-01-01"
END_DATE = "2024-12-31"   # 2 years (low-data but realistic)

dates = pd.date_range(start=START_DATE, end=END_DATE, freq="D")
```

```python
menu_categories = [
    "Rice & Curry",
    "Fried Rice",
    "Kottu",
    "Short Eats",
    "Beverages"
]
```

```python
festival_calendar = {
    "Sinhala_Tamil_New_Year": [
        "2023-04-13", "2023-04-14",
        "2024-04-13", "2024-04-14"
    ],
    "Vesak": [
        "2023-05-05",
        "2024-05-23"
    ]
}
```

```python
rows = []

for date in dates:
    for menu in menu_categories:

        day_of_week = date.day_name()
        is_weekend = 1 if day_of_week in ["Saturday", "Sunday"] else 0
        month = date.month

        # Festival flags
        is_festival = 0
        festival_type = "None"
        for fest, fest_dates in festival_calendar.items():
            if date.strftime("%Y-%m-%d") in fest_dates:
                is_festival = 1
                festival_type = fest

        # Poya day (approximation)
        is_poya_day = 1 if date.day in [14, 15] else 0

        # Weather (Sri Lanka realistic ranges)
        temperature_avg = round(np.random.uniform(26, 33), 1)
        rainfall_mm = round(np.random.exponential(4), 1)
        is_heavy_rain = 1 if rainfall_mm > 20 else 0

        # Proxy data (supplier invoice quantity)
        supplier_invoice_qty = np.random.randint(80, 200)

        # Base demand estimation
        base_demand = supplier_invoice_qty * np.random.uniform(0.7, 0.95)

        if is_weekend:
            base_demand *= 1.2

        if is_festival:
            base_demand *= 1.5

        if is_heavy_rain:
            base_demand *= 0.85

        daily_demand_qty = int(round(base_demand))

        rows.append([
            date,
            day_of_week,
            is_weekend,
            month,
            menu,
            is_festival,
            festival_type,
            is_poya_day,
            temperature_avg,
            rainfall_mm,
            is_heavy_rain,
            supplier_invoice_qty,
            daily_demand_qty
        ])
```

```python
columns = [
    "date",
    "day_of_week",
    "is_weekend",
    "month",
    "menu_category",
    "is_festival",
    "festival_type",
    "is_poya_day",
    "temperature_avg",
    "rainfall_mm",
    "is_heavy_rain",
    "supplier_invoice_qty",
    "daily_demand_qty"  # TARGET
]

df = pd.DataFrame(rows, columns=columns)
```
